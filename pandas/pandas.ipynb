Pandas has two main structures

Series: a labelled one-dimentional array(one coloumn)

DataFrame: a table of labelled coloumns, each a series, sharing a common index.ex: excel spreadsheet

```python
import pandas as pd

s = pd.Series([10,20,30], index = ['a','b','c'])

print(s['b'])
```

```python
df = pd.DataFrame({
    'name': ['Sid','Sai','Likitha','Mohan'],
    'age' : [25,20,22,23],
    'city': ['Hyderabad','Chennai','Bangalore','Mumbai'],
    'score': [90,85,78,65]
})

print(df)

df.to_csv('output.csv', index = False )
```
